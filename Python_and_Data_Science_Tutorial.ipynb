Data Scientist Toolkit Cheatsheet
--
1.** Basic Data Structures**</Br>
Lists: Ordered, mutable collections—e.g., [1, 2, 3].
Use: Store sequences; access with `list[0]`, add with `list.append(x)`.

2. **Control Flow**</Br>
  **For Loops:** Iterate over items—e.g., </Br> `for x in [1, 2, 3]: print(x)`.
  Use: Process sequences step-by-step.

  **If Statements:** Conditional checks—e.g., if x > 0: print("Positive").
  Use: Filter or decide based on conditions.

  **Truthiness:** Non-zero/non-empty is True—e.g., if [1]: print("Not empty").
  Use: Quick checks without explicit comparisons.

3. **Functions & Code Organization**</Br>
  **Functions:** Reusable code blocks—</Br>e.g., `def add(a, b): return a + b`.
  **Use:** Encapsulate logic; call with `add(2, 3)`.

  **List Comprehensions:** Build lists in one line—e.g., `[x * 2 for x in [1, 2, 3]]`.
  Use: Create filtered/transformed lists fast—outputs `[2, 4, 6]`.

  **Object-Oriented Programming (OOP):** Classes for structure—e.g., `class Counter: def __init__(self): self.count = 0`.</Br>
  **Use:** Bundle data and behavior; extend with inheritance like `class DoubleCounter(Counter)`.

4. **Iteration Tools**</Br>
**Iterables:** Anything loopable—e.g., lists, strings—`for x in "abc": print(x)`.
Use: Process items sequentially.

  **Generators:** Lazy iteration—e.g., `def count(): yield 1; yield 2`.
  Use: Yield items one-by-one, saving memory—loop with `for x in count()`.

  enumerate: Pairs indices with items—e.g., `for i, x in enumerate(["a", "b"]): print(i, x)`.
  Use: Get position and value—outputs 0 a, 1 b.

  zip: Combines iterables—e.g., `zip(["a", "b"], [1, 2])`.
  Use: Pair data—outputs `[("a", 1), ("b", 2)]`; unpack with `for x, y in zip()`.

5. **Randomness**</Br>
  **random Module:** Pseudorandom tools—`import random.
random.random(): Float` 0.0 to 1.0—e.g., 0.7421.

  `random.randrange(start, stop[, step])`: Integer in range—e.g., `random.randrange(1, 5) (1-4)`.

  `random.choice(sequence)`: One item—e.g., `random.choice(["a", "b"])` picks "a".

  `random.sample(sequence, k)`: k unique items—e.g., `random.sample([1, 2, 3], 2)`.

  `random.shuffle(list)`: Reorders in place—e.g., `[1, 2, 3] to [3, 1, 2]`.

  Use: Simulate data, pick samples, mix sequences.

6. **Text Processing**</Br>
`Regular Expressions (re): Pattern matching—`import re`.</Br>
  re.search(pattern, string): First match—e.g., re.search(r"\d+", "123abc")`.

  `re.findall(pattern, string)`: All matches—e.g., `re.findall(r"\d+", "1 2 3")` gives `["1", "2", "3"]`.

  `re.sub(pattern, repl, string)`: Replace—e.g., `re.sub(r"\d+", "X", "1abc")` gives "`Xabc`".

  Use: Extract, clean, or transform text—e.g., find numbers, fix formats.

7. **Code Clarity**
  **Type Annotations:** Optional type hints</Br> `x: int = 5, def add(a: int, b: int) -> int`.</Br>
  **Use:** Label variables/functions—e.g., `list[str], dict[str, int]`v—for readability and error-checking (with tools like mypy).

##Default Dictionary ´defaultdict

>>A **defaultdict** is like a regular dictionary, except that **when you try to look up a key it doesn’t contain, it first adds a value for it using a zero-argument function you provided when you created it**. In order to use defaultdicts, you have to import them from collections:

>**defaultdicts are useful** when we’re using dictionaries to defaultdict **“collect” results by some key and don’t want to have to check every time to see if the key exists yet**.


**Use Case for Word Count**: Imagine that you’re trying to count the words in a document. An obvious approach is to create a dictionary in which the **keys are words** and the **values are counts**. As you check each word, you can increment its count if it’s already in the dictionary and add it to the dictionary if it’s not:


```python
word_counts = {}
for word in document:
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

#You could also just handle the exception from trying to look up a missing key:
#The “forgiveness is better than permission”

approach and

word_counts = {}
for word in document:
    try:
        word_counts[word] += 1
    except KeyError:
        word_counts[word] = 1

#A third approach is to use get, which behaves gracefully for missing keys:

word_counts = {}
for word in document:
    previous_count = word_counts.get(word, 0)
    word_counts[word] = previous_count + 1

```

In order to use **defaultdicts**, you have to **import them from collections**:

```python
from collections import defaultdict

word_counts = defaultdict(int)          # int() produces 0
for word in document:
    word_counts[word] += 1

```

```python
dd_list = defaultdict(list)             # list() produces an empty list
dd_list[2].append(1)                    # now dd_list contains {2: [1]}

dd_dict = defaultdict(dict)             # dict() produces an empty dict
dd_dict["Joel"]["City"] = "Seattle"     # {"Joel" : {"City": Seattle"}}

dd_pair = defaultdict(lambda: [0, 0])
dd_pair[2][1] = 1                       # now dd_pair contains {2: [0, 1]}

print(dd_list)
print(dd_dict)
print(dd_pair)
```

**defaultdict(list)** creates a defaultdict where the default value for any key is an empty list ([]).

>"*Keys are words, Values are Counts*"

**dd_list[2].append(1)**: This line accesses the value associated with the **key 2**.</br> Since the key doesn't exist yet, defaultdict automatically creates an empty list for the key 2 and then appends 1 to that list.

**dd_dict["Joel"]["City"] = "Seattle"**: This line accesses the value associated with the key **"Joel"**. Because it doesn't exist, a new empty dictionary is created and assigned to "Joel". </br> Then, within that nested dictionary, the key "City" is assigned the value "Seattle".

**defaultdict(lambda: [0, 0])**: Here, the defaultdict is created with a **lambda function** as the factory. This **lambda function returns a new list [0, 0] each time a missing key is accessed**.</br>
**dd_pair[2][1] = 1**: Similar to the previous examples, this accesses the list associated with the key 2.</br> Since it's missing, the lambda function creates the **list [0, 0]**. </br>Then, the element at **index 1** within this list (the second element) is assigned the **value 1**.

**Generators:** Functions that “yield” **values one-by-one** instead of **making a full list upfront**. They’re iterables too, but **“lazy”—producing items only when needed**.






```python
def generate_range(n):
    i = 0
    while i < n:
        yield i   # every call to yield produces a value of the generator
        i += 1

counter_JT = generate_range(10)
for num in counter_JT:
    print(num)
```

```python
def generate_range(n):
    i = 0
    while i < n:
        yield i   # every call to yield produces a value of the generator
        i += 1

for i in generate_range(10):
    print(f"i: {i}")

```


1.   **yield:** Gives one value **for each i**,</Br> -pauses the function </Br> -resumes next time it’s asked for more.

2.   **Lazy:** No [1, 2, 3] in memory—checks just one number at a time.
3. **for i loop** will consume the yielded values one at a time until none are left







The following loop will consume the yielded values one at a time until none are left:



```python
def count_to_three():
    yield 1  # Hands back 1, pauses
    yield 2  # Hands back 2, pauses
    yield 3  # Hands back 3, done

counter = count_to_three()  # Creates a generator
for num in counter:         # Loops over it
    print(num)              # 1, 2, 3
```

**What It Is:** The “internal state” is **the PRNG’s memory**—think of it as **a secret number that decides what comes next**. It’s like the position in a giant, prewritten list of “random” numbers.

**How It Works**: Each call to a random function (e.g., random.randrange(1, 10)) **uses the current state to compute a number, then updates the state for the next call**.













```python
import random

# Fake web data
def get_options():
    return ["soup", "pizza", "salad", "bread", "pasta"]

options = get_options()

# Random numbers
print("Chance:", random.random())  # e.g., 0.6732
print("Range pick:", random.randrange(1, 11, 2))  # 1, 3, 5, 7, or 9

# Choice and sample
pick = random.choice(options)
print("Single pick:", pick)  # e.g., "soup"
sample = random.sample(options, 3)
print("3 picks:", sample)  # e.g., ["pasta", "bread", "salad"]

# Shuffle with seed
random.seed(25)
random.shuffle(options)
print("Shuffled:", options)  # e.g., ["bread", "salad", "pizza", "soup", "pasta"]
```

```python
def mixed(a, b, *args, **kwargs):
    print(f"a: {a}, b: {b}") #These are regular positional arguments.
    print(f"args: {args}") #*args:collects any extra positional arguments into a tuple.
    print(f"kwargs: {kwargs}") #**kwargs: This collects any extra keyword arguments into a dictionary.

mixed(1, 2, 3, 4, x=5, y=6)
# a: 1, b: 2
# args: (3, 4)
# kwargs: {'x': 5, 'y': 6}
```

Trees: Exploring Line Charts with the Bias-Variance Example
--

>Returning to Continuous variables, line charts using **plt.plot()**, emphasizing their strength in showing trends over a continuous variable—like time or complexity. The excerpt provides a concrete example with variance, bias squared, and total error.


Three lines show how variance rises, bias falls, and total error balances as complexity increases.

```python
import matplotlib.pyplot as plt

variance = [1, 2, 4, 8, 16, 32, 64, 128, 256]        # Increasing variance
bias_squared = [256, 128, 64, 32, 16, 8, 4, 2, 1]     # Decreasing bias^2
total_error = [x + y for x, y in zip(variance, bias_squared)]  # Sum of both
xs = [i for i, _ in enumerate(variance)]              # Indices: 0-8

plt.plot(xs, variance, "g-", label="variance")         # Green solid line
plt.plot(xs, bias_squared, "r-.", label="bias^2")      # Red dot-dashed line
plt.plot(xs, total_error, "b:", label="total error")   # Blue dotted line

plt.legend(loc=9)  # Top center legend
plt.xlabel("model complexity")
plt.xticks([])     # No X-axis ticks
plt.title("The Bias-Variance Tradeoff")
plt.show()
```

**Breaking Down the Example:**</Br></Br>
**Data:**</Br>


1.   **variance:**This list stores values representing the variance of a machine learning model. Notice that the values are increasing. Grows exponentially—model’s sensitivity to data.
2.   **bias_squared:** This list stores values representing the squared bias of the model. These values are decreasing. Drops exponentially—model’s error from oversimplification.
3.   **total_error:** This list is calculated using a list comprehension. It adds the corresponding elements of variance and bias_squared to get the total error.Sum via **zip**—shows the tradeoff’s U-shape.
4.   **X-Axis:** This list is created using enumerate to get the indices (0, 1, 2, ...) of the variance list. These indices will be used as the x-axis values in the plot.xs—indices (0-8)—represent increasing complexity (no specific units).
5.   **This line creates the first line plot:**</Br>**`plt.plot(xs, variance, "g-", label="variance")`**:</Br>
   *   `xs:` x-axis values (model complexity).
   *    variance: y-axis values.
   *   `g-`: Format string for a green solid line.
   *  `label="variance"`: Sets the label for the legend.
>**`plt.plot(xs, bias_squared, "r-.", label="bias2")`**</Br>
>**`plt.plot(xs, total,error, "b:", label="total error")`**
    ***Styles:** "`g-`" (green solid), "`r-.`" (red dot-dashed), "`b:`" (blue dotted)—differentiate series.</Br></Br>
**Customizing Elements:**</Br>
6. **Legend:** `plt.legend(loc=9)`

  *  `loc=9=` Paramter positions Legend depending on variable parameter line. top center  (loc=9 of 0-10 positions)

7. **No Ticks:** `plt.xticks([])`
  *  **hides X-axis values** for simplicity (conceptual focus).
8. **Labels:** `plt.xlabel()`, `plt.title()`
  * top center contextualize the tradeoff.
9. **Why Line Charts Here?:**</Br>
Trends: Smooth lines track changes—perfect for continuous shifts (e.g., complexity’s effect).
10. **Multiple Series:** Multiple `plt.plot()` calls compare related metrics on one graph.

**Vector Operations:**</Br>
--
**Addition:** Add vectors **component-wise**-same length required.



```python
v1 = [1, 2, 3]
v2 = [4, 5, 6]
sum_v = [x + y for x, y in zip(v1, v2)]
print(sum_v)  # [5, 7, 9]
```



*   `for x, y in` ...: The loop iterates through these pairs, assigning the first element of each pair to x and the second element to y.
*   `x + y:` Inside the loop, x and y are added together. The resulting sum is used to create a new list (i.e., creates the resultant vector).





**Python lists aren’t vectors:**

*   If **two vectors v and w** are the **same length**, their sum is just the vector whose **first element** is `v[0] + w[0]`, whose second element is `v[1] + w[1]`, and so on.

*   Adding the vectors `v=[1, 2]` and `w=[2, 1]` results in `[1 + 2, 2 + 1]` or `[3, 3]`,

* Implement this by **zip-ing** the vectors together and **using a list comprehension to add the corresponding elements**:



```python
v1 = [1, 2, 3]
v2 = [4, 5, 6]
sum_v = [x + y for x, y in zip(v1, v2)]
print(sum_v)  # [5, 7, 9]

def add(v1, v2):
    """Adds corresponding elements"""
    assert len(v1) == len(v2), "vectors must be the same length"

    return [v1_i + v2_i for v1_i, v2_i in zip(v1, v2)]

assert add([1, 2, 3], [4, 5, 6]) == [5, 7, 9]

```
